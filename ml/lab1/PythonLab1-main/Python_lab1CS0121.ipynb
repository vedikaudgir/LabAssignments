```python
# use of an iterator
# iterator - returns numbers that start with 25
# and each time it is passed, the number gets squared.

class MyInit:
    def __iter__(self):
        self.a = 25
        return self

    def __next__(self):
        x = self.a
        self.a = self.a ** 2
        return x


obj1 = MyInit()
itr = iter(obj1)

print(next(itr))
print(next(itr))
print(next(itr))


my_list = [1, 2, 3, 4, 5]

print(my_list)

for i in range(len(my_list)):
    my_list[i] = my_list[i] ** 2

print(my_list)
```

Group the cars by number of cylinder, and finding the average cty mpg for each group.

```python
MpgByCyl = []

for c in cylinders:
    summpg = 0
    typecount = 0

    for d in mpg:
        if d['cyl'] == c:
            summpg += int(d['cty'])
            typecount += 1

    MpgByCyl.append((c, summpg / typecount))

MpgByCyl.sort(key=lambda x: x[0])

MpgByCyl
```

Return the unique values for the class types in our dataset using set.

```python
vehicleclass = set(d['class'] for d in mpg)

print(vehicleclass)
```

Find average miles per gallon in the highway for each class of vehicles

```python
MileageByClass = []

for t in vehicleclass:
    sumpg = 0
    classCount = 0

    for d in mpg:
        if d['class'] == t:
            sumpg += int(d['hwy'])
            classCount += 1

    MileageByClass.append((t, sumpg / classCount))

MileageByClass.sort(key=lambda x: x[1])

MileageByClass
```
